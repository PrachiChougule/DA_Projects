### **Data Visualization**

```python
sns.histplot(data=AI_job, x='experience_level')
plt.title('Job Distribution by Experience Level')
plt.xlabel('Experience Level')
plt.ylabel('No. of Jobs')
```

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=AI_job,
    x='years_experience',
    y='salary_usd'
)

plt.title('Salary vs Years of Experience')
plt.xlabel('Years of Experience')
plt.ylabel('Salary (USD)')
plt.show()
```

```python
top_5_job_roles = AI_job['job_title'].value_counts().head(5)

ax = sns.barplot(top_5_job_roles)

plt.xticks(rotation=90)
plt.xlabel("Job Role")
plt.ylabel("Number of Jobs")
plt.title("Top 5 Job Roles")

for container in ax.containers:
    ax.bar_label(container)

plt.show()
```

```python
plt.figure(figsize=(10, 6))
plt.pie(AI_job['employment_type'].value_counts(), labels=AI_job['employment_type'].value_counts().index,autopct='%1.1f%%')
plt.title('Employment Type Distribution')
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(
    data=AI_job,
    x='company_size',
    y='salary_usd'
)

plt.title('Salary by Company Size')
plt.xlabel('Company Size')
plt.ylabel('Salary (USD)')
plt.show()
```

```python
top_5_locations = AI_job.groupby('company_location').size().sort_values(ascending=False)[:5]

sns.countplot(data=AI_job, y='company_location', order=top_5_locations.index)
plt.title('Top 5 Hiring Locations')
```

```python
plt.figure(figsize=(10, 6))

sns.boxplot(
    data=AI_job,
    x='industry',
    y='benefits_score'
)

plt.title('Benefits Score by Industry')
plt.xlabel('Industry')
plt.ylabel('Benefits Score')
plt.xticks(rotation=45)
plt.show()
```

```python
remote_exp = pd.DataFrame({
    0: AI_job[AI_job['remote_ratio'] == 0]['experience_level'].value_counts(),
    50: AI_job[AI_job['remote_ratio'] == 50]['experience_level'].value_counts(),
    100: AI_job[AI_job['remote_ratio'] == 100]['experience_level'].value_counts()
})

sns.heatmap(remote_exp, annot=True, fmt='d',cmap='Greens')

plt.title("Remote Work by Experience Level")
plt.xlabel("Remote Ratio")
plt.ylabel("Experience Level")
plt.show()
```

```python
sns.histplot(AI_job['years_experience'], bins=10)

plt.title("Distribution of Years of Experience")
plt.xlabel("Years of Experience")
plt.ylabel("Number of Jobs")
plt.show()  
```

```python
ax = sns.barplot(
    data=AI_job,
    x='job_title',
    y='years_experience'
    
)

plt.xticks(rotation=90)
plt.title("Years of Experience by Job Role")
plt.xlabel("Job Role")
plt.ylabel("Years of Experience")
plt.show()
```

### **Key Insights**

- Salary usually increases with more experience.
- Some AI and data jobs have more openings than others.
- Large companies usually offer higher salaries than smaller companies.
- Remote, hybrid, and on-site jobs are available at different experience levels.
- Job opportunities and benefits are different across locations and industries
